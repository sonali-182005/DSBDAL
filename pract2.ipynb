```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
data = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Karan'],
    'Marks': [85, 90, np.nan, 35, 400],
    'Attendance': [90, 95, 85, np.nan, 80]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original Dataset:")
print(df)
```

```python
# Fill missing marks with mean
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
```

```python

# Fill missing attendance with mean
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].mean())
```

```python
print("\nDataset After Handling Missing Values:")
print(df)
```

```python
Q1 = df['Marks'].quantile(0.25)
Q3 = df['Marks'].quantile(0.75)
IQR = Q3 - Q1
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
```

```python
print("\nLower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
# Detect outliers
outliers = df[(df['Marks'] < lower_limit) | (df['Marks'] > upper_limit)]
```

```python
print("\nOutliers:")
print(outliers)
```

```python
df['Log_Marks'] = np.log(df['Marks'])
```

```python
print("\nLog Transformation:")
print(df[['Marks', 'Log_Marks']])
```

```python
minmax = MinMaxScaler()

df['Normalized_Marks'] = minmax.fit_transform(df[['Marks']])
```

```python
print("\nNormalized Data:")
print(df[['Marks', 'Normalized_Marks']])
```

```python
scaler = StandardScaler()
df['Scaled_Marks'] = scaler.fit_transform(df[['Marks']])
```

```python
print("\nScaled Data:")
print(df[['Marks', 'Scaled_Marks']])
```
